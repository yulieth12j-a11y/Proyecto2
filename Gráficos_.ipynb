```python
import matplotlib.pyplot as plt
import numpy as np

dias = ["Lunes", "Martes", "Miercoles", "Jueves", "Viernes"]
horas = [1, 2, 1.5, 2.5, 3]

plt.plot(dias, horas, marker="o", color="orange")
plt.title("Horas de entrenamiento durante la semana")
plt.xlabel("Dias")
plt.ylabel("Horas")
plt.grid()
plt.show()
```
